# Demo 1: de rekening delen

Werkt prima, tot je er iets raars in gooit.

```python
bedrag = float(input("Totaalbedrag: "))
personen = int(input("Aantal personen: "))

per_persoon = bedrag / personen

print(f"Elk {per_persoon} euro")
```
