# Demo: code die draait, er goed uitziet, en toch fout is

Verzonnen functies (`lijst.add()`, `random.pick()`) crashen meteen en zijn dus onschuldig.
Je merkt ze op het moment zelf. Code die stilletjes het verkeerde antwoord geeft, merkt
niemand op, en die gaat mee tot in productie.

Een functie zoals een AI-assistent ze zou voorstellen bij de vraag *"schrijf een functie
die het gemiddelde van een lijst getallen berekent"*.

Ze is niet verzonnen. Ze is niet lelijk. Ze werkt zelfs. Meestal.

```python
def gemiddelde(getallen):
    totaal = 0
    for getal in getallen:
        totaal = totaal + getal
    return totaal / len(getallen)
```

### Test 1: het gelukkige geval

Dit werkt. Hier stoppen de meeste mensen met testen. Dat is de fout.

```python
print("Test 1:", gemiddelde([10, 20, 30]))
```

### Test 2: de lege lijst

```python
print("Test 2:", gemiddelde([]))
```

`ZeroDivisionError: division by zero`

Dit soort fouten zouden we moeten zien aankomen: `len([])` is 0, en delen door 0 kan niet.
Dit is "redeneren over code", waar les 2 helemaal over gaat.

### Test 3: de belangrijkste

```python
print("Test 3:", gemiddelde(["10", "20", "30"]))
```

Dus toch een crash. Goed. Maar bekijk nu deze variant:

```python
def gemiddelde_variant(getallen):
    totaal = ""
    for getal in getallen:
        totaal = totaal + getal
    return len(totaal) / len(getallen)


print("Test 3b:", gemiddelde_variant(["10", "20", "30"]))
```

`2.0`

GEEN crash. Een keurig getal. Volledig fout. Het gemiddelde van 10, 20 en 30 is 20, niet 2.

Waarom krijgen we 2 terug? Dít is waarom "het draait" nooit betekent "het klopt".
