# Data ophalen met PokéAPI

Begin in de browser: <https://pokeapi.co/api/v2/pokemon/pikachu/>. Een GET-request vraagt data op; de response bevat een statuscode en JSON.

Installeer vooraf `requests` met `uv pip install requests` in de bestaande module-venv. Kies diezelfde omgeving als notebookkernel. Voer de cellen in volgorde uit.

Bronnen: [PokéAPI](https://pokeapi.co/docs/v2#pokemon) en [Requests](https://requests.readthedocs.io/en/latest/user/quickstart/).

### 1. Een GET-request

200 betekent gelukt. `raise_for_status()` maakt een exception van een HTTP-fout. `timeout=10` begrenst het wachten bij netwerkstilte, niet de totale downloadtijd.

```python
import requests

url = "https://pokeapi.co/api/v2/pokemon/pikachu/"
response = requests.get(url, timeout=10)
print(f"Statuscode: {response.status_code}")
response.raise_for_status()
```

### 2. JSON wordt een Python-dictionary

`response` is het volledige antwoord; `pokemon` bevat de omgezette gegevens. Welke sleutels herken je uit de browser?

```python
pokemon = response.json()
print(type(pokemon))  # <class 'dict'>
print(pokemon["name"])
print(pokemon["height"])
```

### 3. Bekende Python-code op echte gegevens

Gebruik dictionaries en lussen uit les 4 en 5. Lengte: decimeter naar meter. Gewicht: hectogram naar kilogram. Haal één keer op en hergebruik `pokemon` terwijl je de uitvoer aanpast.

```python
print(f"Naam: {pokemon['name'].capitalize()}")
print(f"Lengte: {pokemon['height'] / 10:.1f} m")  # API: decimeter
print(f"Gewicht: {pokemon['weight'] / 10:.1f} kg")  # API: hectogram
print("Types:")
for type_info in pokemon["types"]:
    print(f"- {type_info['type']['name']}")
```

### 4. Later bij try/except: fouten gericht opvangen

Voer deze cel pas uit na het blok over `try`/`except`. Test eerst `bestaatniet`, daarna `pikachu`. `else` draait alleen als de aanvraag gelukt is. Je eigen tikfouten blijven zichtbaar.

```python
try:
    response = requests.get(
        "https://pokeapi.co/api/v2/pokemon/bestaatniet/", timeout=10
    )
    response.raise_for_status()
except requests.exceptions.HTTPError:
    if response.status_code == 404:
        print("Die Pokémon bestaat niet. Controleer de naam.")
    else:
        print(f"De API meldt een HTTP-fout: {response.status_code}.")
except requests.exceptions.Timeout:
    print("De API reageert te traag. Probeer later opnieuw.")
except requests.exceptions.ConnectionError:
    print("Geen verbinding met PokéAPI. Controleer je internet.")
else:
    pokemon = response.json()
    print(pokemon["name"])
```
